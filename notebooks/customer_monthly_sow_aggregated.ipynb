```python
%sql
CREATE OR REPLACE TABLE synchrony.analytics.customer_monthly_sow_final AS

SELECT
    *,

    CASE
        WHEN Total_MetroMart_Spend > 0
        THEN HSIC_Spend / Total_MetroMart_Spend
        ELSE 0
    END AS HSIC_SoW

FROM synchrony.analytics.customer_monthly_sow;
```

```python
%sql
select * from synchrony.analytics.customer_monthly_sow_final
```

```python
%sql
SELECT *

FROM synchrony.analytics.customer_monthly_sow_final

ORDER BY Customer_ID, Month

LIMIT 50;
```
