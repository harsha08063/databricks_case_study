```python
%sql
CREATE OR REPLACE TABLE synchrony.analytics.customer_monthly_sow_final AS
SELECT
    *,
    CASE
        WHEN Total_MetroMart_Spend <= 0
            THEN NULL
        WHEN HSIC_Spend < 0
          OR HSIC_Spend > Total_MetroMart_Spend
            THEN NULL
        ELSE HSIC_Spend / Total_MetroMart_Spend
    END AS HSIC_SoW,

    CASE
        WHEN Total_MetroMart_Spend <= 0
            THEN 'INVALID_NON_POSITIVE_SPEND'
        WHEN HSIC_Spend < 0
          OR HSIC_Spend > Total_MetroMart_Spend
            THEN 'INVALID_SOW_RANGE'
        ELSE 'VALID'
    END AS SoW_Status

FROM synchrony.analytics.customer_monthly_sow;
```

```python
%sql
select * from synchrony.analytics.customer_monthly_sow_final
```

```python
%sql
SELECT *

FROM synchrony.analytics.customer_monthly_sow_final

ORDER BY Customer_ID, Month

LIMIT 50;
```

```python
%sql
SELECT
    SoW_Status,
    COUNT(*) AS Rows,
    MIN(HSIC_SoW) AS Min_SoW,
    MAX(HSIC_SoW) AS Max_SoW
FROM synchrony.analytics.customer_monthly_sow_final
GROUP BY SoW_Status
ORDER BY SoW_Status;
```
