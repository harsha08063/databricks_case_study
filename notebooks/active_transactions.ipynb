```python
%sql
CREATE OR REPLACE TABLE synchrony.analytics.active_transactions AS

SELECT *

FROM synchrony.analytics.transaction_net_sales

WHERE Transaction_Date >= Credit_Card_Open_Date

AND (
    Credit_Card_Closed_Date IS NULL
    OR Transaction_Date <= Credit_Card_Closed_Date
);
```
