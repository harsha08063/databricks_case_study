```python
%sql
CREATE OR REPLACE TABLE synchrony.analytics.transaction_net_sales AS

SELECT
    *,
    
    CASE
        WHEN Transaction_Type = 'Return'
            THEN -ABS(Transaction_Amount)
        ELSE ABS(Transaction_Amount)
    END AS Net_Amount,

    CASE
        WHEN Transaction_Type = 'Return'
            THEN 0
        ELSE COALESCE(Number_of_Transactions, 0)
    END AS Net_Transactions

FROM synchrony.analytics.transaction_enriched;
```

```python
%sql
SELECT
    Transaction_Type,
    COUNT(*) AS Rows,
    SUM(Net_Amount) AS Net_Sales
FROM synchrony.analytics.transaction_net_sales
GROUP BY Transaction_Type;
```

```python
%sql
SELECT
    Transaction_Type,
    COUNT(*) AS Rows,
    SUM(Transaction_Amount) AS Gross_Amount,
    SUM(Net_Amount) AS Net_Amount,
    SUM(Number_of_Transactions) AS Original_Transactions,
    SUM(Net_Transactions) AS Net_Transactions
FROM synchrony.analytics.transaction_net_sales
GROUP BY Transaction_Type
ORDER BY Transaction_Type;
```
