```python
%sql
CREATE OR REPLACE TABLE synchrony.analytics.report_category_summary AS
SELECT
    Category,

    COUNT(DISTINCT Customer_ID) AS Customer_Count,

    SUM(FY2025_Total_Spend) AS FY2025_Total_Spend,
    SUM(FY2026_Total_Spend) AS FY2026_Total_Spend,

    SUM(FY2025_HSIC_Spend) AS FY2025_HSIC_Spend,
    SUM(FY2026_HSIC_Spend) AS FY2026_HSIC_Spend,

    CASE
        WHEN SUM(FY2025_Total_Spend) > 0
        THEN SUM(FY2025_HSIC_Spend)
             / SUM(FY2025_Total_Spend)
        ELSE NULL
    END AS FY2025_SoW,

    CASE
        WHEN SUM(FY2026_Total_Spend) > 0
        THEN SUM(FY2026_HSIC_Spend)
             / SUM(FY2026_Total_Spend)
        ELSE NULL
    END AS FY2026_SoW,

    SUM(FY2026_Total_Spend)
      - SUM(FY2025_Total_Spend) AS Total_Spend_Change,

    SUM(FY2026_HSIC_Spend)
      - SUM(FY2025_HSIC_Spend) AS HSIC_Spend_Change,

    (
        CASE
            WHEN SUM(FY2026_Total_Spend) > 0
            THEN SUM(FY2026_HSIC_Spend)
                 / SUM(FY2026_Total_Spend)
            ELSE NULL
        END
    )
    -
    (
        CASE
            WHEN SUM(FY2025_Total_Spend) > 0
            THEN SUM(FY2025_HSIC_Spend)
                 / SUM(FY2025_Total_Spend)
            ELSE NULL
        END
    ) AS SoW_Change

FROM synchrony.analytics.report_category_customer

GROUP BY Category;
```

```python
%sql
SELECT *
FROM synchrony.analytics.report_category_summary
ORDER BY SoW_Change ASC;
```
