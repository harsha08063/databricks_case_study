```python
%sql
SELECT

    Fiscal_Year,

    SUM(Total_MetroMart_Spend) AS Total_Spend,

    SUM(HSIC_Spend) AS HSIC_Spend,

    SUM(Other_Payment_Spend) AS Other_Payment_Spend,

    SUM(HSIC_Spend)
        / SUM(Total_MetroMart_Spend) AS Overall_SoW

FROM synchrony.analytics.customer_monthly_sow_final

GROUP BY Fiscal_Year

ORDER BY Fiscal_Year;
```

```python
%sql
SELECT

    Customer_ID,

    Fiscal_Year,

    SUM(Total_MetroMart_Spend) AS Total_Spend,

    SUM(HSIC_Spend) AS HSIC_Spend,

    SUM(HSIC_Spend)
        / NULLIF(SUM(Total_MetroMart_Spend),0) AS SoW

FROM synchrony.analytics.customer_monthly_sow_final

GROUP BY
    Customer_ID,
    Fiscal_Year

ORDER BY
    Customer_ID,
    Fiscal_Year;
```
