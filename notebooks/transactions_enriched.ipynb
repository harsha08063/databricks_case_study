```python
%sql
CREATE OR REPLACE TABLE synchrony.analytics.transaction_enriched AS

SELECT
    t.Customer_ID,
    t.Transaction_ID,
    CAST(t.Transaction_Date AS DATE) AS Transaction_Date,

    t.Category_Code,
    cat.Category,

    t.Transaction_Type,
    t.Transaction_Amount,

    t.Payment_Code,
    p.Payment_Method,

    t.Number_of_Transactions,

    cust.Age,
    cust.Gender,
    cust.Membership_Type,

    CAST(cust.Credit_Card_Open_Date AS DATE)
        AS Credit_Card_Open_Date,

    CAST(cust.Credit_Card_Closed_Date AS DATE)
        AS Credit_Card_Closed_Date,

    cust.Credit_Card_Limit,
    cust.Credit_Card_APR

FROM synchrony.analytics.transacations_data t

LEFT JOIN synchrony.analytics.customer_data cust
    ON t.Customer_ID = cust.Customer_ID

LEFT JOIN synchrony.analytics.payment_code p
    ON t.Payment_Code = p.Payment_Code

LEFT JOIN synchrony.analytics.category_code cat
    ON t.Category_Code = cat.Category_Code;
```

```python
%sql
select * from synchrony.analytics.transaction_enriched limit 10
```
