```python
%sql
CREATE OR REPLACE TABLE synchrony.analytics.report_customer_segments AS
SELECT
    SoW_Movement,
    HSIC_Movement,
    Total_Spend_Movement,

    COUNT(*) AS Customer_Count,

    SUM(FY2025_HSIC_Spend) AS FY2025_HSIC_Spend,
    SUM(FY2026_HSIC_Spend) AS FY2026_HSIC_Spend,

    SUM(FY2025_Total_Spend) AS FY2025_Total_Spend,
    SUM(FY2026_Total_Spend) AS FY2026_Total_Spend,

    SUM(HSIC_Spend_Change) AS HSIC_Spend_Change,
    SUM(Total_Spend_Change) AS Total_Spend_Change

FROM synchrony.analytics.customer_movement_analysis

WHERE SoW_Movement <> 'NOT_COMPARABLE'

GROUP BY
    SoW_Movement,
    HSIC_Movement,
    Total_Spend_Movement;
```

```python
%sql
SELECT *
FROM synchrony.analytics.report_customer_segments
ORDER BY Customer_Count DESC;
```
